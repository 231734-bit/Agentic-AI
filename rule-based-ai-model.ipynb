```python
def weather_advice(temperature_c, humidity_percent, wind_speed_kmh, air_quality_index):
    """Rule-based weather advisor using simple IF/THEN logic."""

    advice = []

    # Rule 1: Temperature
    if temperature_c >= 35:
        advice.append("Extreme heat: stay hydrated and avoid the sun.")
    elif temperature_c <= 5:
        advice.append("Cold weather: wear a warm jacket.")

    # Rule 2: Humidity
    if humidity_percent >= 80:
        advice.append("High humidity: expect possible rain, carry an umbrella.")

    # Rule 3: Wind speed - NEW RULE
    if wind_speed_kmh >= 40:
        advice.append("Strong wind warning: be careful outdoors.")

    # Rule 4: Air quality - NEW RULE
    if air_quality_index >= 150:
        advice.append("Poor air quality: avoid outdoor activities if possible.")

    # If no rule was triggered
    if not advice:
        advice.append("Conditions look moderate, no special precautions needed.")

    return advice


if __name__ == "__main__":
    sample_days =[
        (42, 75, 30, 100),     # Extreme heat
        (4, 65, 20, 80),       # Cold
        (27, 88, 25, 120),     # High humidity
        (32, 55, 48, 140),     # Strong wind
        (20, 45, 15, 175)
]

    for temp, humidity, wind, air_quality in sample_days:
        result = weather_advice(temp, humidity, wind, air_quality)

        print(
            f"Temp={temp}C, "
            f"Humidity={humidity}%, "
            f"Wind={wind} km/h, "
            f"Air Quality Index={air_quality} "
            f"-> {result}"
        )
```

```python
# Simple Data-Driven Spam Classifier

spam_keywords = ["free", "winner", "urgent", "click here", "prize"]

# Threshold: if 2 or more spam keywords are found, classify as spam
threshold = 4


def classify_email(email_text):
    email_text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in email_text:
            count += 1

    if count >= threshold:
        return "Spam"
    else:
        return "Legitimate"


# Like facial recognition or medical diagnosis, this classifier assigns an input to a fixed category.

# Six test emails with correct labels
test_emails = [
    ("Congratulations! You are a winner. Claim your free prize now!", "Spam"),
    ("Urgent! Click here to receive your free prize.", "Spam"),
    ("You have won a free gift. Click here now!", "Spam"),
    ("Hi, can you send me the notes from today's class?", "Legitimate"),
    ("Your meeting is scheduled for tomorrow at 10 AM.", "Legitimate"),
    ("Please review the project report and send me your feedback.", "Legitimate")
]


# Test the classifier
correct = 0

for email, correct_label in test_emails:
    predicted_label = classify_email(email)

    print("Email:", email)
    print("Correct Label:", correct_label)
    print("Predicted Label:", predicted_label)
    print()

    if predicted_label == correct_label:
        correct += 1


# Calculate accuracy
accuracy = (correct / len(test_emails)) * 100

print("Correct Predictions:", correct)
print("Total Emails:", len(test_emails))
print("Accuracy:", accuracy, "%")
```
